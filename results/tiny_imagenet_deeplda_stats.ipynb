# TinyImageNet final test accuracy (mean ± stdev)

Loads `results/tiny_imagenet_deeplda.json` and reports final test accuracy across three runs for both heads.


```python
from pathlib import Path
import json
import statistics as stats

data_path = Path("tiny_imagenet_deeplda.json")
with data_path.open() as f:
    data = json.load(f)

def mean_std(values):
    mean = stats.mean(values)
    stdev = stats.stdev(values) if len(values) > 1 else 0.0
    return 100*mean, 100*stdev

for head_name in sorted(data["results"]):
    finals = [r["final_test_acc"] for r in data["results"][head_name]]
    mean, stdev = mean_std(finals)
    print(f"{head_name} final_test_acc: {mean:.2f} ± {stdev:.2f}")
```
