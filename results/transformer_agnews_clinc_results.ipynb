# Transformer results: mean ± stdev test accuracy

Rows = heads, columns = datasets.


```python
from pathlib import Path
import json
import statistics as stats
import pandas as pd

data_path = Path("transformer_agnews_clinc_two_heads.json")
with data_path.open() as f:
    data = json.load(f)

def mean_std(values):
    mean = stats.mean(values)
    stdev = stats.stdev(values) if len(values) > 1 else 0.0
    return mean, stdev

datasets = sorted(data["results"].keys())
heads = sorted(next(iter(data["results"].values())).keys())

table = []
for head in heads:
    row = {"head": head}
    for dataset in datasets:
        runs = data["results"][dataset][head]
        finals = [r["final_test_acc"] for r in runs]
        mean, stdev = mean_std(finals)
        row[dataset] = f"{mean:.4f} ± {stdev:.4f}"
    table.append(row)

df = pd.DataFrame(table).set_index("head").loc[heads, datasets]
df
```
